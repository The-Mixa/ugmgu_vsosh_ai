## Аргументы сортировки

```python
sorted(iterable, /, *, key=None, reverse=False)
items.sort(*, key=None, reverse=False)
```

`sorted(...)` возвращает **новый список**, не меняя исходные данные. `items.sort(...)` меняет сам список и возвращает `None`.

- `reverse=False` задаёт направление: при `False` порядок возрастающий, при `True` — убывающий.
- `key=None` — функция, которая получает один элемент и возвращает значение для сравнения. Примеры: `key=len`, `key=str.lower`, `key=lambda x: abs(x)`.

Ключ может быть кортежем. Компоненты сравниваются слева направо: `key=lambda row: (-row[2], row[0])` ставит большие значения `row[2]` первыми, а при равенстве — имена по алфавиту.

Сортировка стабильна: равные ключи сохраняют исходный порядок. Если нужен явный порядок при равенстве, добавьте его в ключ.

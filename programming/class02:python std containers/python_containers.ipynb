## 1. `str` — строки

### Как устроены строки

- Строка — **неизменяемая (immutable)** упорядоченная последовательность символов Unicode.
- С Python 3.3 действует «гибкое представление строк» (PEP 393): интерпретатор сам выбирает
  компактный формат хранения (1, 2 или 4 байта на символ) в зависимости от самого «тяжёлого»
  символа в строке. Строка из ASCII-символов занимает меньше памяти, чем строка с эмодзи.
- Раз строки неизменяемы, **любая операция, которая выглядит как «изменение», на деле создаёт
  новый объект** в памяти — исходная строка не трогается.
- Короткие «простые» строки (похожие на идентификаторы) интерпретатор может **интернировать**
  (interning) — кешировать так, что два одинаковых литерала указывают на один и тот же объект.
- Строки **хешируемы** — именно поэтому их можно использовать как ключи словаря или элементы множества.

### Свойства
| Свойство | Значение |
|---|---|
| Упорядоченность | да, порядок символов важен |
| Изменяемость | нет (immutable) |
| Хешируемость | да |
| Индексация/срезы | да |
| Повторяющиеся элементы | допустимы |
| Типы элементов | только символы (строки длины 1) |

```python
s1 = "python"
s2 = "python"
# функция id(obj) показывает внутренний адрес python на  obj в памяти компьютера
print(s1 is s2, id(s1) == id(s2))

s = "python"
print(id(s))
s = s + "!"   # выглядит как "изменение", но это НОВЫЙ объект
print(id(s))
```

### Операции над строками

```python
s = "python"

# индексация и срезы
print(s[0], s[-1])          # первый и последний символ
print(s[1:4])                # срез
print(s[::-1])                # разворот строки
print(s[::2])                 # каждый второй символ

# конкатенация и повторение
print("py" + "thon")
print("ab" * 3)

# проверка вхождения и сравнение
print("th" in s, "z" in s)
print("abc" < "abd")          # лексикографическое сравнение

# длина, итерация, распаковка
print(len(s))
for ch in s[:3]:
    print(ch, end=" ")
print()
a, b, c = s[:3]
print(a, b, c)
```

### Форматирование строк

```python
name, score = "Михаил", 100

print(f"{name} набрал {score} баллов")   # f-строка (рекомендуемый способ)
print(f"{3.14159:.2f}")                  # форматирование числа
```

### Основные методы в деле

```python
s = "  Привет, Мир!  "
print(repr(s.strip()))
print(s.strip().lower())
print(s.strip().upper())
print(s.strip().replace("Мир", "Python"))
print("a,b,,c".split(","))
print("-".join(["2026", "09", "24"]))
print("42".isdigit(), "abc".isalpha(), "abc123".isalnum())
print("hello world".title())
print("hello".startswith("he"), "hello".endswith("lo"))
print("hello".find("l"), "hello".rfind('l'), "hello".index("l"))
print("hello".count("l"))
```

| Длина | `len(s)` | `len("abc")` → `3` | O(1) |
| Итерация | `for ch in s` | — | O(n) |

## 2. `list` — списки

### Как устроены списки

- Список — **изменяемый (mutable)** динамический массив: под капотом это непрерывный блок памяти,
  хранящий **указатели на объекты**, а не сами объекты.
- Чтобы `append` работал быстро, CPython выделяет память "с запасом" (over-allocation): при
  нехватке места массив пересоздаётся с некоторым запасом ячеек, поэтому добавление в конец —
  O(1) **амортизированно**, а не всегда мгновенно.
- Вставка/удаление **в начале или середине** требует сдвига всех последующих элементов — O(n).
- Список **не хешируем** — его нельзя использовать как ключ словаря или элемент множества.
- Присваивание `b = a` не копирует список — `a` и `b` указывают на один и тот же объект
  (нужно `a.copy()`, `list(a)` или срез `a[:]` для поверхностной копии).

### Свойства
| Свойство | Значение |
|---|---|
| Упорядоченность | да |
| Изменяемость | да (mutable) |
| Хешируемость | нет |
| Индексация/срезы | да |
| Повторяющиеся элементы | допустимы |
| Типы элементов | любые, можно смешивать |

```python
a = [1, 2, 3]
b = a               # b — это ссылка на тот же объект!
b.append(4)
print(a, b, a is b)

c = a.copy()         # поверхностная копия
c.append(100)
print(a, c)
```

### Операции над списками

```python
lst = [10, 20, 30, 40, 50]

# индексация и срезы
print(lst[0], lst[-1])
print(lst[1:4])
print(lst[::-1])

# изменение через индекс и срез
lst[0] = 99
print(lst)
lst[1:3] = [1000, 2000, 3000]     # срез можно заменить на кусок другой длины
print(lst)
del lst[0]                         # удаление по индексу
print(lst)

# конкатенация и повторение (создают НОВЫЙ список)
x = [1, 2] + [3, 4]
y = [0] * 3
print(x, y)

# += изменяет список на месте (эквивалент extend), в отличие от +
z = [1, 2]
z += [3, 4]
print(z)

# членство, сравнение, агрегаты
nums = [5, 3, 8, 1]
print(3 in nums)
print([1, 2] < [1, 3])             # лексикографическое сравнение
print(len(nums), min(nums), max(nums), sum(nums))

# сортировка: sorted() возвращает новый список, .sort() сортирует на месте
print(sorted(nums))
nums.sort(reverse=True)
print(nums)

# списковые включения (list comprehension)
squares = [x**2 for x in range(6) if x % 2 == 0]
print(squares)

# вложенные списки и поверхностное копирование — частая ловушка
matrix = [[0, 0], [0, 0]]
bad_copy = matrix.copy()
bad_copy[0][0] = 1                  # поверхностная копия НЕ копирует вложенные списки
print(matrix)
```

### Шпаргалка: методы `list`

| Метод | Что делает | Пример | Результат |
|---|---|---|---|
| `l.append(x)` | добавить элемент в конец | `[1].append(2)` | `[1, 2]` |
| `l.extend(it)` | добавить все элементы итерируемого объекта | `[1].extend([2,3])` | `[1, 2, 3]` |
| `l.insert(i, x)` | вставить по индексу | `[1,3].insert(1,2)` | `[1, 2, 3]` |
| `l.remove(x)` | удалить первое вхождение значения | `[1,2,1].remove(1)` | `[2, 1]` |
| `l.pop([i])` | удалить и вернуть элемент (по умолчанию последний) | `[1,2,3].pop()` | `3` |
| `l.clear()` | удалить все элементы | `[1,2].clear()` | `[]` |
| `l.index(x)` | индекс первого вхождения | `[1,2,3].index(2)` | `1` |
| `l.count(x)` | сколько раз встречается | `[1,1,2].count(1)` | `2` |
| `l.sort(key=, reverse=)` | сортировать на месте | `[3,1].sort()` | `[1, 3]` |
| `l.reverse()` | развернуть на месте | `[1,2].reverse()` | `[2, 1]` |
| `l.copy()` | поверхностная копия | `[1,2].copy()` | `[1, 2]` |

### Шпаргалка: операции над `list`

| Операция | Синтаксис | Пример | Сложность |
|---|---|---|---|
| Индексация | `l[i]` | `[1,2,3][0]` → `1` | O(1) |
| Срез | `l[a:b:c]` | `[1,2,3,4][1:3]` → `[2,3]` | O(k) |
| Присваивание по индексу | `l[i] = x` | — | O(1) |
| `append` в конец | `l.append(x)` | — | O(1) амортизированно |
| Вставка/удаление в начале/середине | `l.insert(i,x)` / `del l[i]` | — | O(n) |
| Конкатенация | `l1 + l2` | `[1]+[2]` → `[1,2]` | O(n+m) |
| Членство | `x in l` | — | O(n) |
| Сортировка | `sorted(l)` / `l.sort()` | — | O(n·log n) |
| Длина | `len(l)` | — | O(1) |

## 3. `tuple` — кортежи

### Как устроены кортежи

- Кортеж — **неизменяемый (immutable)** аналог списка: под капотом это тоже массив указателей
  на объекты, но **фиксированного размера**, выделяемый ровно под нужное число элементов
  (без запаса, в отличие от списка).
- Раз размер и содержимое не меняются, кортеж **хешируем** — но только если хешируемы все его
  элементы (кортеж, содержащий список, хешировать нельзя).
- Из-за неизменяемости и отсутствия over-allocation кортежи немного компактнее и быстрее
  создаются, чем списки той же длины — их удобно использовать для «связок» данных, которые
  не должны меняться (например, координаты точки, запись базы данных).
- Кортеж из одного элемента требует запятой: `(1,)`, а не `(1)` (это просто число в скобках).

### Свойства
| Свойство | Значение |
|---|---|
| Упорядоченность | да |
| Изменяемость | нет (immutable) |
| Хешируемость | да, если хешируемы все элементы |
| Индексация/срезы | да |
| Повторяющиеся элементы | допустимы |
| Типы элементов | любые, можно смешивать |

```python
point = (3, 4)
print(hash(point))                 # кортеж из хешируемых элементов — хешируем
d = {point: "точка A"}             # поэтому кортеж можно использовать как ключ словаря
print(d)

bad = ([1, 2], 3)
try:
    hash(bad)                       # а вот кортеж со списком внутри — нельзя
except TypeError as e:
    print("Ошибка:", e)
```

### Операции над кортежами

```python
t = (10, 20, 30, 40)

# индексация и срезы — как у списка
print(t[0], t[-1], t[1:3])

# распаковка (unpacking) — очень частый паттерн для кортежей
x, y, z, w = t
print(x, y, z, w)
first, *middle, last = t
print(first, middle, last)

# упаковка нескольких значений в кортеж
def min_max(nums):
    return min(nums), max(nums)     # возвращает кортеж (min, max)

lo, hi = min_max([5, 1, 9, 3])
print(lo, hi)

# конкатенация и повторение — создают новый кортеж
print((1, 2) + (3, 4))
print((0,) * 3)

# сравнение и сортировка — как у списков, лексикографически
print((1, 2) < (1, 3))
print(sorted([(2, "b"), (1, "z"), (1, "a")]))   # сортировка по элементам кортежа по очереди

# конвертация между list и tuple
lst = list(t)
lst.append(50)
t2 = tuple(lst)
print(t2)
```

### Шпаргалка: методы `tuple`

У кортежа всего два метода — из-за неизменяемости методов, меняющих содержимое, попросту нет.

| Метод | Что делает | Пример | Результат |
|---|---|---|---|
| `t.count(x)` | сколько раз встречается значение | `(1,1,2).count(1)` | `2` |
| `t.index(x)` | индекс первого вхождения | `(5,6,7).index(6)` | `1` |

### Шпаргалка: операции над `tuple`

| Операция | Синтаксис | Пример | Сложность |
|---|---|---|---|
| Индексация | `t[i]` | `(1,2,3)[0]` → `1` | O(1) |
| Срез | `t[a:b:c]` | `(1,2,3,4)[1:3]` → `(2,3)` | O(k) |
| Конкатенация | `t1 + t2` | `(1,)+(2,)` → `(1,2)` | O(n+m) |
| Повторение | `t * n` | `(1,)*3` → `(1,1,1)` | O(n·k) |
| Членство | `x in t` | — | O(n) |
| Распаковка | `a, b = t` | — | O(n) |
| Длина | `len(t)` | — | O(1) |
| Хеш (если применимо) | `hash(t)` | — | O(n) |

## 4. `dict` — словари

### Как устроены словари

- Словарь — **изменяемая** коллекция пар «ключ → значение», основанная на **хеш-таблице**.
- Ключ хешируется функцией `hash()`, по хешу вычисляется позиция в таблице — поэтому доступ,
  вставка и удаление по ключу работают в среднем за **O(1)**, а не за O(n), как поиск в списке.
- Именно поэтому **ключи должны быть хешируемыми** (и, как следствие, обычно неизменяемыми):
  подходят `str`, `int`, `tuple` (из хешируемых элементов), но не подходят `list`, `dict`, `set`.
- С Python 3.6 (и официально с 3.7) словарь устроен как «компактный dict»: отдельно хранится
  плотный массив записей в порядке добавления и отдельная разреженная хеш-таблица с индексами
  в этот массив. Побочный эффект — **словарь сохраняет порядок добавления ключей**, это гарантия
  языка, а не случайность.
- Значения могут быть любыми объектами, в том числе изменяемыми.

### Свойства
| Свойство | Значение |
|---|---|
| Упорядоченность | да, по порядку добавления ключей (с 3.7) |
| Изменяемость | да (mutable) |
| Хешируемость ключей | обязательна |
| Уникальность ключей | да, значения могут повторяться |
| Доступ по ключу | в среднем O(1) |

```python
d = {"a": 1, "b": 2}
try:
    d[[1, 2]] = "x"        # список нельзя использовать как ключ — он не хешируем
except TypeError as e:
    print("Ошибка:", e)

d2 = {}
d2["z"] = 1
d2["a"] = 2
d2["m"] = 3
print(list(d2.keys()))       # порядок добавления сохраняется, а не алфавитный
```

### Операции над словарями

```python
student = {"name": "Михаил", "faculty": "ИИ", "year": 1}

# доступ, безопасный доступ, изменение, удаление
print(student["name"])
print(student.get("age", "не указано"))    # get не бросает ошибку, если ключа нет
student["year"] = 2
student["age"] = 18
del student["age"]
print(student)

# проверка наличия ключа (`in` проверяет именно ключи, не значения)
print("faculty" in student, "Михаил" in student)

# итерация: по ключам, значениям, парам
for key in student:
    print(key, "->", student[key])
for key, value in student.items():
    print(f"{key}: {value}")

# объединение словарей
defaults = {"lang": "ru", "year": 1}
override = {"year": 2, "active": True}
merged_old = {**defaults, **override}      # через распаковку
merged_new = defaults | override           # оператор | (Python 3.9+)
print(merged_old)
print(merged_new)

d3 = dict(defaults)
d3.update(override)                         # update изменяет на месте
print(d3)

# словарные включения (dict comprehension)
squares = {x: x**2 for x in range(5)}
print(squares)

# сравнение словарей — по содержимому, порядок не важен
print({"a": 1, "b": 2} == {"b": 2, "a": 1})
```

## 5. `set` — множества

### Как устроены множества

- Множество — **изменяемая** неупорядоченная коллекция **уникальных** хешируемых элементов,
  тоже построенная на хеш-таблице (как словарь, только без значений — по сути «словарь, у
  которого есть только ключи»).
- Благодаря хеш-таблице операции добавления, удаления и проверки на вхождение в среднем
  выполняются за **O(1)** — именно поэтому `x in some_set` намного быстрее, чем `x in some_list`
  на больших объёмах данных.
- В отличие от `dict`, у `set` **нет** компактного представления с сохранением порядка — порядок
  элементов не гарантирован и может выглядеть «случайным».
- Есть неизменяемый «близнец» — `frozenset`: как `set`, но без методов изменения, а значит
  хешируем и годится как ключ словаря или элемент другого множества.
- Элементы множества обязаны быть хешируемыми (нельзя положить список внутрь множества).

### Свойства
| Свойство | Значение |
|---|---|
| Упорядоченность | нет |
| Изменяемость | да (`set`) / нет (`frozenset`) |
| Хешируемость элементов | обязательна |
| Уникальность элементов | да, дубликаты автоматически схлопываются |
| Проверка вхождения | в среднем O(1) |

```python
nums = {1, 2, 2, 3, 3, 3}      # дубликаты схлопываются сами
print(nums)

try:
    bad = {[1, 2], 3}          # список внутри множества — нельзя, он не хешируем
except TypeError as e:
    print("Ошибка:", e)

fs = frozenset([1, 2, 3])
d = {fs: "неизменяемое множество как ключ"}   # а frozenset — можно
print(d)
```

### Операции над множествами

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

# теоретико-множественные операции (создают новое множество)
print(a | b)            # объединение
print(a & b)             # пересечение
print(a - b)              # разность: есть в a, но нет в b
print(a ^ b)               # симметрическая разность: есть ровно в одном из множеств

# те же операции методами (полезно, если аргумент — не множество, а любой iterable)
print(a.union([10, 11]))
print(a.intersection([2, 3, 100]))

# операции "на месте" — изменяют исходное множество
c = {1, 2, 3}
c |= {4, 5}     # эквивалент c.update({4, 5})
print(c)
c &= {1, 4, 5}
print(c)

# отношения между множествами
print({1, 2} <= {1, 2, 3})       # подмножество
print({1, 2, 3} >= {1, 2})        # надмножество
print({1, 2}.isdisjoint({3, 4}))   # нет общих элементов

# членство и включения
print(3 in a)
unique_letters = {ch for ch in "abracadabra"}   # включение множества (set comprehension)
print(unique_letters)

# типичное применение: быстрая дедупликация и пересечение списков
list1 = [1, 2, 2, 3, 4]
print(list(set(list1)))          # уникальные элементы без сохранения порядка
```

### Шпаргалка: методы `set`

| Метод | Что делает | Пример | Результат |
|---|---|---|---|
| `s.add(x)` | добавить элемент | `{1}.add(2)` | `{1, 2}` |
| `s.remove(x)` | удалить элемент (ошибка, если нет) | `{1,2}.remove(2)` | `{1}` |
| `s.discard(x)` | удалить элемент (без ошибки, если нет) | `{1}.discard(5)` | `{1}` |

| `s.union(other)` | объединение (то же, что \| ) | `{1}.union({2})` | `{1, 2}` |
| `s.intersection(other)` | пересечение (то же, что &) | `{1,2}.intersection({2,3})` | `{2}` |
| `s.difference(other)` | разность (то же, что -) | `{1,2}.difference({2})` | `{1}` |
| `s.symmetric_difference(other)` | симметрическая разность (то же, что ^) | `{1,2}.symmetric_difference({2,3})` | `{1, 3}` |
| `s.update(other)` | объединить на месте (то же, что \|=) | `{1}.update({2})` | `{1, 2}` |
| `s.intersection_update(other)` | пересечь на месте (то же, что &=) | — | — |
| `s.difference_update(other)` | вычесть на месте (то же, что -=) | — | — |
| `s.symmetric_difference_update(other)` | симм. разность на месте (то же, что ^=) | — | — |
| `s.isdisjoint(other)` | нет общих элементов? | `{1}.isdisjoint({2})` | `True` |

| `s.issuperset(other)` | является надмножеством? (то же, что >=) | `{1,2}.issuperset({1})` | `True` |

## Базовый уровень

### Строки (`str`)

**Решение 1.** Срез с шагом -1 проходит строку от конца к началу.

```python
def reverse_string(s):
    return s[::-1]

print(reverse_string("python"))
```

**Решение 2.** Убираем пробелы и приводим к одному регистру, затем сравниваем строку с её разворотом.

```python
def is_palindromic(s):
    cleaned = s.lower().replace(" ", "")
    return cleaned == cleaned[::-1]

print(is_palindromic("А роза упала на лапу Азора"))
print(is_palindromic("hello"))
```

**Решение 3.** Множество гласных даёт проверку `in` за O(1), а генератор считает совпадения.

```python
def count_vowels(s):
    vowels = set("aeiouAEIOU")
    return sum(1 for ch in s if ch in vowels)

print(count_vowels("Hello, World!"))
```

**Решение 4.** Метод `str.replace` создаёт новую строку с заменённой подстрокой.

```python
def replace_spaces(s):
    return s.replace(" ", "_")

print(replace_spaces("hello world foo"))
```

**Решение 5.** Разбиваем по пробелу, у каждого слова вручную делаем первую букву заглавной, склеиваем обратно.

```python
def manual_title(s):
    words = s.split(" ")
    capitalized = [w[0].upper() + w[1:] if w else w for w in words]
    return " ".join(capitalized)

print(manual_title("hello world"))
```

### Списки (`list`)

**Решение 6.** Встроенные `sum()` и `len()` дают сумму и позволяют вычислить среднее.

```python
def sum_and_average(nums):
    total = sum(nums)
    average = total / len(nums)
    return total, average

print(sum_and_average([1, 2, 3, 4, 5]))
```

**Решение 7.** Множество `seen` хранит уже встреченные значения — проверка вхождения O(1), порядок сохраняется за счёт обычного списка-результата.

```python
def dedupe_preserve_order(items):
    seen = set()
    result = []
    for item in items:
        if item not in seen:
            seen.add(item)
            result.append(item)
    return result

print(dedupe_preserve_order([3, 1, 3, 2, 1, 4, 2]))
```

**Решение 8.** Проходим список один раз, попутно отслеживая наибольшее и второе по величине значение.

```python
def second_largest(nums):
    largest = second = float("-inf")
    for n in nums:
        if n > largest:
            largest, second = n, largest
        elif n > second:
            second = n
    return second

print(second_largest([5, 1, 9, 3, 7]))
```

**Решение 9.** Два указателя (`left`, `right`) сходятся к середине, меняя элементы местами.

```python
def manual_reverse(lst):
    lst = lst.copy()
    left, right = 0, len(lst) - 1
    while left < right:
        lst[left], lst[right] = lst[right], lst[left]
        left += 1
        right -= 1
    return lst

print(manual_reverse([1, 2, 3, 4, 5]))
```

**Решение 10.** Два списковых включения с разными условиями на чётность.

```python
def split_even_odd(nums):
    evens = [n for n in nums if n % 2 == 0]
    odds = [n for n in nums if n % 2 != 0]
    return evens, odds

print(split_even_odd([1, 2, 3, 4, 5, 6]))
```

### Кортежи (`tuple`)

**Решение 11.** `max()` с параметром `key` находит кортеж, у которого второй элемент (возраст) максимален.

```python
def find_oldest(people):
    return max(people, key=lambda person: person[1])

people = [("Анна", 20), ("Борис", 35), ("Вера", 28)]
print(find_oldest(people))
```

**Решение 12.** Распаковываем кортеж в переменные и собираем новый кортеж в обратном порядке.

```python
def swap_tuple(t):
    a, b = t
    return (b, a)

print(swap_tuple((1, 2)))
```

**Решение 13.** Метод `count` — один из двух методов кортежа — считает вхождения значения.

```python
t = (3, 1, 3, 5, 3, 2)
print(t.count(3))
```

**Решение 14.** Распаковка кортежа сразу даёт именованные переменные `x, y, z` для формулы.

```python
import math

def vector_length(point):
    x, y, z = point
    return math.sqrt(x**2 + y**2 + z**2)

print(vector_length((3, 4, 0)))
```

**Решение 15.** Конкатенация `+` объединяет кортежи, `sorted()` возвращает отсортированный список.

```python
def merge_sorted(t1, t2):
    return sorted(t1 + t2)

print(merge_sorted((3, 1, 4), (1, 5, 9)))
```

### Словари (`dict`)

**Решение 16.** `dict.get(word, 0)` позволяет наращивать счётчик, даже если слово встретилось впервые.

```python
def word_frequency(words):
    freq = {}
    for w in words:
        freq[w] = freq.get(w, 0) + 1
    return freq

print(word_frequency(["a", "b", "a", "c", "b", "a"]))
```

**Решение 17.** Словарное включение проходит по парам `.items()` и меняет местами ключ и значение.

```python
def invert_dict(d):
    return {v: k for k, v in d.items()}

print(invert_dict({"a": 1, "b": 2, "c": 3}))
```

**Решение 18.** `max()` с `key=prices.get` ищет ключ, для которого значение словаря максимально.

```python
def most_expensive(prices):
    return max(prices, key=prices.get)

print(most_expensive({"хлеб": 60, "молоко": 90, "сыр": 450}))
```

**Решение 19.** Копируем первый словарь, затем для каждого ключа из второго прибавляем значение через `get(key, 0)`.

```python
def merge_sum(d1, d2):
    result = dict(d1)
    for key, value in d2.items():
        result[key] = result.get(key, 0) + value
    return result

print(merge_sum({"a": 1, "b": 2}, {"b": 3, "c": 4}))
```

**Решение 20.** Словарное включение с условием фильтрует пары по значению.

```python
def filter_by_age(people, min_age):
    return {name: age for name, age in people.items() if age >= min_age}

print(filter_by_age({"Аня": 17, "Боря": 22, "Вера": 19}, 18))
```

### Множества (`set`)

**Решение 21.** Множество автоматически схлопывает дубликаты; `list()` возвращает элементы обратно списком.

```python
def unique_elements(nums):
    return list(set(nums))

print(unique_elements([1, 2, 2, 3, 1, 4]))
```

**Решение 22.** Оператор `&` находит элементы, присутствующие в обоих множествах.

```python
def common_elements(a, b):
    return list(set(a) & set(b))

print(common_elements([1, 2, 3, 4], [3, 4, 5]))
```

**Решение 23.** `issubset` (или оператор `<=`) проверяет, что все элементы `A` есть в `B`.

**Решение 24.** Оператор `-` (разность множеств) оставляет элементы, которых нет во втором множестве.

```python
def only_in_first(a, b):
    return list(set(a) - set(b))

print(only_in_first([1, 2, 3, 4], [3, 4, 5]))
```

**Решение 25.** Если длина множества символов меньше длины строки — значит, какой-то символ повторился.

```python
def has_duplicate_chars(s):
    return len(set(s)) != len(s)

print(has_duplicate_chars("hello"))
print(has_duplicate_chars("world"))
```

## Средний уровень

**Решение 26.** Регулярное выражение выделяет слова без пунктуации, словарь считает частоты, `sorted` по значению с `reverse=True` выбирает топ.

```python
import re

def top_words(text, top_n=3):
    words = re.findall(r"[а-яa-z]+", text.lower())
    freq = {}
    for w in words:
        freq[w] = freq.get(w, 0) + 1
    return sorted(freq.items(), key=lambda pair: pair[1], reverse=True)[:top_n]

text = "Python is great. Python is easy. I love Python and I love easy code."
print(top_words(text))
```

**Решение 27.** Множества сводят обе операции к готовым инструментам: `&` — общие элементы, `^` — те, что ровно в одном списке.

```python
def compare_student_lists(list1, list2):
    set1, set2 = set(list1), set(list2)
    common = set1 & set2
    only_one = set1 ^ set2
    return common, only_one

group_a = ["Аня", "Боря", "Вера", "Аня"]
group_b = ["Вера", "Гриша", "Боря"]
print(compare_student_lists(group_a, group_b))
```

**Решение 28.** Словарь накапливает стоимость по товару, затем `sorted()` по значению строит рейтинг по убыванию.

```python
def total_cost_per_product(orders):
    totals = {}
    for product, price, qty in orders:
        totals[product] = totals.get(product, 0) + price * qty
    ranked = sorted(totals.items(), key=lambda pair: pair[1], reverse=True)
    return totals, ranked

orders = [
    ("Хлеб", 60, 3),
    ("Молоко", 90, 2),
    ("Хлеб", 60, 1),
    ("Сыр", 450, 1),
]
print(total_cost_per_product(orders))
```

**Решение 29.** Извлекаем слова регулярным выражением, множественное включение фильтрует по длине, `tuple(sorted(...))` убирает повторы и сортирует.

```python
import re

def long_unique_words(sentence):
    words = re.findall(r"[а-яa-z]+", sentence.lower())
    long_words = {w for w in words if len(w) > 4}
    return tuple(sorted(long_words))

sentence = "Python is a powerful, flexible and popular programming language."
print(long_unique_words(sentence))
```

**Решение 30.** `dict.setdefault(key, []).append(value)` — стандартный приём группировки без проверки «есть ли уже такой ключ».

```python
def group_by(pairs):
    groups = {}
    for key, value in pairs:
        groups.setdefault(key, []).append(value)
    return groups

data = [("fruits", "apple"), ("veg", "carrot"), ("fruits", "banana"), ("veg", "potato")]
print(group_by(data))
```

## Сложный уровень

**Решение 31.** Для каждого документа берём множество уникальных слов (чтобы не дублировать индекс при повторе слова в одном тексте) и добавляем номер документа в множество по каждому слову.

```python
def build_inverted_index(documents):
    index = {}
    for doc_id, text in enumerate(documents):
        for word in set(text.lower().split()):
            index.setdefault(word, set()).add(doc_id)
    return index

docs = [
    "python is great",
    "java is verbose",
    "python is easy and great",
]
index = build_inverted_index(docs)
for word in sorted(index):
    print(word, "->", index[word])
```

**Решение 32.** Приводим обе строки к спискам букв без пробелов/пунктуации, считаем частоту каждой буквы словарём и сравниваем два словаря целиком.

```python
def is_anagram(s1, s2):
    def normalize(s):
        return [ch for ch in s.lower() if ch.isalpha()]

    def char_counts(chars):
        counts = {}
        for ch in chars:
            counts[ch] = counts.get(ch, 0) + 1
        return counts

    return char_counts(normalize(s1)) == char_counts(normalize(s2))

print(is_anagram("Listen", "Silent"))
print(is_anagram("Hello", "World"))
```

**Решение 33.** Множество `seen` хранит значения ключа, которые уже встречались; список-результат пополняется, только если значение встречено впервые — так сохраняется порядок.

```python
def dedupe_by_key(records, key):
    seen = set()
    result = []
    for record in records:
        value = record[key]
        if value not in seen:
            seen.add(value)
            result.append(record)
    return result

people = [
    {"id": 1, "name": "Аня"},
    {"id": 2, "name": "Боря"},
    {"id": 1, "name": "Аня (дубликат)"},
    {"id": 3, "name": "Вера"},
]
print(dedupe_by_key(people, "id"))
```

**Решение 34.** Словарь смежности хранит множество соседей для каждой вершины (граф неориентированный — ребро добавляется в обе стороны). Соседи соседей дают вершины на расстоянии два шага; объединяем оба множества и убираем саму стартовую вершину.

```python
def build_adjacency(edges):
    adjacency = {}
    for a, b in edges:
        adjacency.setdefault(a, set()).add(b)
        adjacency.setdefault(b, set()).add(a)
    return adjacency

def reachable_within_two_steps(adjacency, start):
    one_step = adjacency.get(start, set())
    two_steps = set()
    for neighbor in one_step:
        two_steps |= adjacency.get(neighbor, set())
    return (one_step | two_steps) - {start}

edges = [("A", "B"), ("B", "C"), ("C", "D"), ("A", "E")]
adjacency = build_adjacency(edges)
print(adjacency)
print(reachable_within_two_steps(adjacency, "A"))
```

**Решение 35.** Шифрование сдвигает код каждой буквы по модулю 26 в пределах своего алфавита (регистр учитывается отдельной базовой точкой отсчёта). Взлом перебирает все 26 сдвигов, расшифровывает текст каждым из них и сравнивает частоту букв результата с эталонной частотой английского языка — побеждает сдвиг с наименьшим суммарным отклонением.

```python
import string

ALPHABET = string.ascii_lowercase

def caesar_encrypt(text, shift):
    result = []
    for ch in text:
        if ch.isalpha():
            base = ord('A') if ch.isupper() else ord('a')
            result.append(chr((ord(ch) - base + shift) % 26 + base))
        else:
            result.append(ch)
    return "".join(result)

def caesar_decrypt(text, shift):
    return caesar_encrypt(text, -shift)

# эталонные частоты букв английского языка (в процентах, приблизительно)
ENGLISH_FREQ = {
    'e': 12.7, 't': 9.1, 'a': 8.2, 'o': 7.5, 'i': 7.0, 'n': 6.7,
    's': 6.3, 'h': 6.1, 'r': 6.0, 'd': 4.3, 'l': 4.0, 'c': 2.8,
    'u': 2.8, 'm': 2.4, 'w': 2.4, 'f': 2.2, 'g': 2.0, 'y': 2.0,
    'p': 1.9, 'b': 1.5, 'v': 1.0, 'k': 0.8, 'j': 0.15, 'x': 0.15,
    'q': 0.10, 'z': 0.07,
}

def letter_counts(text):
    counts = {letter: 0 for letter in ALPHABET}
    for ch in text.lower():
        if ch.isalpha():
            counts[ch] += 1
    return counts

def score_shift(text, shift):
    decrypted = caesar_decrypt(text, shift).lower()
    counts = letter_counts(decrypted)
    total = sum(counts.values())
    if total == 0:
        return 0
    score = 0
    for letter, expected_pct in ENGLISH_FREQ.items():
        observed_pct = counts[letter] / total * 100
        score -= abs(observed_pct - expected_pct)
    return score

def crack_caesar(ciphertext):
    return max(range(26), key=lambda shift: score_shift(ciphertext, shift))

original = ("Python is a general purpose programming language that lets you work quickly "
            "and integrate systems more effectively. It emphasizes code readability with "
            "the use of significant indentation. Many developers around the world use "
            "python every single day for data science, web development, automation and "
            "artificial intelligence research.")
encrypted = caesar_encrypt(original, 7)
print("Зашифровано:", encrypted[:80], "...")

guessed_shift = crack_caesar(encrypted)
print("Предполагаемый сдвиг:", guessed_shift)
print("Расшифровано:", caesar_decrypt(encrypted, guessed_shift)[:80], "...")
```
